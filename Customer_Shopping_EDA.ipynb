```python
df['invoice_date'] = pd.to_datetime(df['invoice_date'], errors='coerce', dayfirst=True)

df['month_year'] =df ['invoice_date'].dt.to_period('M').astype(str)

df['total_price'] =df['quantity'] * df['price']
monthly_sales = df.groupby('month_year')['total_price'].sum().reset_index()

fig = px.line(monthly_sales, x='month_year', y='total_price',
              title="Sales Trend Over Time",
              labels={'month_year': 'Month-Year', 'total_price': 'Total Sales Revenue'})

fig.show()
```
